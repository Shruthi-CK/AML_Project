```python
# ============================
# LSTM Implementation with Full Feature Engineering
# ============================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.metrics import mean_squared_error, r2_score

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

import mlflow

# ----------------------------
# Global config
# ----------------------------
np.random.seed(42)
tf.random.set_seed(42)

SEQUENCE_LENGTH = 5  # number of time steps in each LSTM sequence

# ============================
# 0. MLflow global setup
# ============================
mlflow.set_tracking_uri("http://127.0.0.1:5000")
mlflow.set_experiment("AgroFood_CO2_Emissions")

# ============================
# 1. Load fixed train/test splits
# ============================
df_train = pd.read_csv("agro_co2_train.csv")
df_test = pd.read_csv("agro_co2_test.csv")

print("Train CSV shape:", df_train.shape)
print("Test  CSV shape:", df_test.shape)

TRAIN_END_YEAR = int(df_train["Year"].max())
print("Train end year (from CSV):", TRAIN_END_YEAR)

# Combine back for time-based feature engineering (lags/rolling)
df_all = pd.concat([df_train, df_test], ignore_index=True)
df_all = df_all.sort_values(["Area", "Year"]).reset_index(drop=True)

# ============================
# 2. Target engineering (log emissions)
# ============================
if "log_total_emission" not in df_all.columns:
    emission_min = df_all["total_emission"].min()
    shift_amount = 1 - emission_min if emission_min <= 0 else 0
    df_all["total_emission_shifted"] = df_all["total_emission"] + shift_amount
    df_all["log_total_emission"] = np.log(df_all["total_emission_shifted"])

target_col = "log_total_emission"

# ============================
# 3. Define base feature buckets
# ============================

# Land use drivers
land_use_features = [
    "Savanna fires",
    "Forest fires",
    "Forestland",
    "Net Forest conversion",
    "Drained organic soils (CO2)",
    "Fires in organic soils",
    "Fires in humid tropical forests",
]

# Agriculture & livestock drivers
agriculture_livestock_features = [
    "Crop Residues",
    "Rice Cultivation",
    "Manure applied to Soils",
    "Manure left on Pasture",
    "Manure Management",
    "Fertilizers Manufacturing",
    "IPPU",
]

# Energy & supply chain drivers
energy_use_features = [
    "On-farm Electricity Use",
    "Food Processing",
    "Food Packaging",
    "Food Transport",
    "Food Retail",
    "Food Household Consumption",
    "Agrifood Systems Waste Disposal",
    "On-farm energy use",
    "Pesticides Manufacturing",
]

# Demographics
population_features = [
    "Rural population",
    "Urban population",
    "Total Population - Male",
    "Total Population - Female",
]

# Climate + time
temperature_features = ["Average Temperature °C"]
time_features = ["Year"]

categorical_cols = ["Area"]

base_numerical_cols = (
    land_use_features
    + agriculture_livestock_features
    + energy_use_features
    + population_features
    + temperature_features
    + time_features
)

# ============================
# 4. Add lag & rolling features (per Area)
# ============================

def add_time_features(df, target_col, group_col="Area"):
    df = df.sort_values([group_col, "Year"]).copy()

    # 1-year and 2-year lags of log emissions
    df["lag1_log_emission"] = df.groupby(group_col)[target_col].shift(1)
    df["lag2_log_emission"] = df.groupby(group_col)[target_col].shift(2)

    # 3-year rolling mean of log emissions (includes current year)
    df["roll3_log_emission"] = (
        df.groupby(group_col)[target_col]
          .rolling(window=3, min_periods=1)
          .mean()
          .reset_index(level=0, drop=True)
    )

    # Simple year-on-year change in log emissions
    df["d_log_emission"] = df[target_col] - df["lag1_log_emission"]

    # Fill NaNs in engineered features per Area (no leakage across countries)
    lag_cols = ["lag1_log_emission", "lag2_log_emission",
                "roll3_log_emission", "d_log_emission"]

    for col in lag_cols:
        df[col] = (
            df.groupby(group_col)[col]
              .apply(lambda s: s.fillna(method="ffill").fillna(method="bfill"))
        )

    return df

df_all = add_time_features(df_all, target_col=target_col, group_col="Area")

# New engineered numerical columns
engineered_cols = [
    "lag1_log_emission",
    "lag2_log_emission",
    "roll3_log_emission",
    "d_log_emission",
]

numerical_cols = base_numerical_cols + engineered_cols

# Columns we will feed to preprocessor
driver_features = categorical_cols + numerical_cols

# Sanity check
missing_features = [c for c in driver_features if c not in df_all.columns]
if missing_features:
    raise ValueError(f"Missing driver features in df_all: {missing_features}")

# ============================
# 5. Re-split into train/test AFTER feature engineering
# ============================
df_train_fe = df_all[df_all["Year"] <= TRAIN_END_YEAR].copy()
df_test_fe  = df_all[df_all["Year"] > TRAIN_END_YEAR].copy()

print("Train FE shape:", df_train_fe.shape)
print("Test  FE shape:", df_test_fe.shape)

# ============================
# 6. Prepare data for LSTM
# ============================
df_train_lstm = df_train_fe[driver_features + [target_col]].copy()
df_test_lstm  = df_test_fe[driver_features + [target_col]].copy()

# Sort by Area, Year for temporal order
df_train_lstm = df_train_lstm.sort_values(by=["Area", "Year"]).reset_index(drop=True)
df_test_lstm  = df_test_lstm.sort_values(by=["Area", "Year"]).reset_index(drop=True)

# Group key for sequences (per country)
df_train_lstm["group_key"] = df_train_lstm["Area"]
df_test_lstm["group_key"]  = df_test_lstm["Area"]

# Preprocessor: fit ONLY on training features (no leakage)
preprocessor_lstm = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_cols),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols),
    ]
)

X_train_raw = df_train_lstm[categorical_cols + numerical_cols]
X_test_raw  = df_test_lstm[categorical_cols + numerical_cols]

# Fit on train, transform train and test
X_train_proc = preprocessor_lstm.fit_transform(X_train_raw)
X_test_proc  = preprocessor_lstm.transform(X_test_raw)

# Densify if sparse
if hasattr(X_train_proc, "toarray"):
    X_train_proc = X_train_proc.toarray()
if hasattr(X_test_proc, "toarray"):
    X_test_proc = X_test_proc.toarray()

# Get feature names
try:
    feature_names_lstm = preprocessor_lstm.get_feature_names_out()
except AttributeError:
    feature_names_lstm = [f"feat_{i}" for i in range(X_train_proc.shape[1])]

if len(feature_names_lstm) != X_train_proc.shape[1]:
    print(
        f"Warning: len(feature_names_lstm)={len(feature_names_lstm)} "
        f"!= n_features={X_train_proc.shape[1]}, regenerating generic names."
    )
    feature_names_lstm = [f"feat_{i}" for i in range(X_train_proc.shape[1])]

print("X_train_proc shape:", X_train_proc.shape)
print("n feature names   :", len(feature_names_lstm))

# Build feature DataFrames
train_features_df = pd.DataFrame(X_train_proc, columns=feature_names_lstm)
test_features_df  = pd.DataFrame(X_test_proc,  columns=feature_names_lstm)

# Attach target + group_key
train_data = train_features_df.copy()
train_data[target_col] = df_train_lstm[target_col].values
train_data["group_key"] = df_train_lstm["group_key"].values

test_data = test_features_df.copy()
test_data[target_col] = df_test_lstm[target_col].values
test_data["group_key"] = df_test_lstm["group_key"].values

# ============================
# 7. Sequence creation helper
# ============================
def create_sequences(data, target_col, group_col, seq_len):
    """
    Creates 3D sequences for LSTM input.

    Args:
        data: DataFrame with features, target, and group key, sorted by time
        target_col: Name of target column
        group_col: Column to group by (e.g., country)
        seq_len: Length of each sequence

    Returns:
        X_seq: np.array of shape (n_samples, seq_len, n_features)
        y_seq: np.array of shape (n_samples,)
    """
    X_seq, y_seq = [], []

    feature_cols = [c for c in data.columns if c not in [target_col, group_col]]

    for group_name, group_data in data.groupby(group_col):
        features = group_data[feature_cols].values
        target = group_data[target_col].values

        if len(features) >= seq_len:
            for i in range(len(features) - seq_len):
                X_seq.append(features[i : i + seq_len])
                y_seq.append(target[i + seq_len])

    return np.array(X_seq, dtype=np.float32), np.array(y_seq, dtype=np.float32)

# ============================
# 10. LSTM hyperparameter search via MLflow
# ============================

def build_lstm_model(seq_length, n_features, units_1, units_2, dropout):
    """Builds a parametric LSTM model."""
    model = Sequential([
        LSTM(units_1, activation="relu",
             input_shape=(seq_length, n_features),
             return_sequences=True),
        Dropout(dropout),
        LSTM(units_2, activation="relu"),
        Dropout(dropout),
        Dense(1)
    ])
    model.compile(optimizer="adam", loss="mse")
    return model


# -------------------------------
# Hyperparameter grid for LSTM
# -------------------------------
param_grid = {
    "units_1": [32, 64],
    "units_2": [16, 32],
    "dropout": [0.1, 0.2],
    "sequence_length": [4, 5, 6],
    "batch_size": [32, 64]
}

import itertools
param_combinations = list(itertools.product(
    param_grid["units_1"],
    param_grid["units_2"],
    param_grid["dropout"],
    param_grid["sequence_length"],
    param_grid["batch_size"]
))


print(f"\n🔍 Total LSTM configs to test: {len(param_combinations)}")
print("Starting LSTM hyperparameter tuning...\n")


best_score = np.inf
best_config = None
best_run_id = None


# Loop through all combinations
for (u1, u2, drop, seq_len, batch) in param_combinations:

    print(f"\n=== Training LSTM config: u1={u1}, u2={u2}, drop={drop}, "
          f"seq_len={seq_len}, batch={batch} ===")

    # Recreate sequences for this sequence length
    X_train_seq_h, y_train_seq_h = create_sequences(
        train_data,
        target_col=target_col,
        group_col="group_key",
        seq_len=seq_len
    )
    X_test_seq_h, y_test_seq_h = create_sequences(
        test_data,
        target_col=target_col,
        group_col="group_key",
        seq_len=seq_len
    )

    # Skip if too few sequences
    if X_train_seq_h.shape[0] < 10 or X_test_seq_h.shape[0] < 10:
        print("Skipping: not enough sequences.")
        continue

    n_features_h = X_train_seq_h.shape[2]

    with mlflow.start_run(run_name="LSTM_Hyperparam_Sweep", nested=True) as run:

        # Build model for this configuration
        model_h = build_lstm_model(
            seq_length=seq_len,
            n_features=n_features_h,
            units_1=u1,
            units_2=u2,
            dropout=drop
        )

        early_stop = EarlyStopping(
            monitor="val_loss",
            patience=3,
            restore_best_weights=True
        )

        # Train
        history_h = model_h.fit(
            X_train_seq_h, y_train_seq_h,
            epochs=25,
            batch_size=batch,
            validation_data=(X_test_seq_h, y_test_seq_h),
            callbacks=[early_stop],
            verbose=0
        )

        # Predictions
        y_pred_test_h = model_h.predict(X_test_seq_h, verbose=0).flatten()
        rmse_test_h = np.sqrt(mean_squared_error(y_test_seq_h, y_pred_test_h))
        r2_test_h = r2_score(y_test_seq_h, y_pred_test_h)

        # Log metrics
        mlflow.log_metric("rmse_test", rmse_test_h)
        mlflow.log_metric("r2_test", r2_test_h)

        # Log params
        mlflow.log_param("units_1", u1)
        mlflow.log_param("units_2", u2)
        mlflow.log_param("dropout", drop)
        mlflow.log_param("sequence_length", seq_len)
        mlflow.log_param("batch_size", batch)
        mlflow.log_param("epochs_planned", 25)
        mlflow.log_param("train_end_year", TRAIN_END_YEAR)

        print(f"→ Test RMSE = {rmse_test_h:.4f}, R² = {r2_test_h:.4f}")
        print(f"Logged run: {run.info.run_id}")

        # Track best config
        if rmse_test_h < best_score:
            best_score = rmse_test_h
            best_config = (u1, u2, drop, seq_len, batch)
            best_run_id = run.info.run_id


print("\n===========================")
print("      Best LSTM Config     ")
print("===========================")
print(f"Best RMSE: {best_score:.4f}")
print(f"Best hyperparameters:")
print(f" - units_1 = {best_config[0]}")
print(f" - units_2 = {best_config[1]}")
print(f" - dropout = {best_config[2]}")
print(f" - sequence_length = {best_config[3]}")
print(f" - batch_size = {best_config[4]}")
print(f"Best MLflow run ID = {best_run_id}")


# ============================
# 8. Build train/test sequences
# ============================
print("\nGenerating sequences for LSTM...")

X_train_seq, y_train_seq = create_sequences(
    train_data, target_col=target_col, group_col="group_key", seq_len=SEQUENCE_LENGTH
)
X_test_seq, y_test_seq = create_sequences(
    test_data, target_col=target_col, group_col="group_key", seq_len=SEQUENCE_LENGTH
)

print(f"\n✓ X_train_seq shape: {X_train_seq.shape} (n_samples, seq_len, n_features)")
print(f"✓ y_train_seq shape: {y_train_seq.shape}")
print(f"✓ X_test_seq shape: {X_test_seq.shape}")
print(f"✓ y_test_seq shape: {y_test_seq.shape}")

assert X_train_seq.ndim == 3, f"ERROR: X_train_seq must be 3D, got {X_train_seq.ndim}D"
assert y_train_seq.ndim == 1, f"ERROR: y_train_seq must be 1D, got {y_train_seq.ndim}D"

# ============================
# 9. LSTM model + MLflow run
# ============================
if X_train_seq.shape[0] > 0 and X_test_seq.shape[0] > 0:
    n_features = X_train_seq.shape[2]

    with mlflow.start_run(run_name="LSTM_Full_Features_Lags"):
        print("\n=== Training LSTM ===")

        model = Sequential([
            LSTM(64, activation="relu",
                 input_shape=(SEQUENCE_LENGTH, n_features),
                 return_sequences=True),
            Dropout(0.2),
            LSTM(32, activation="relu"),
            Dropout(0.2),
            Dense(1)
        ])

        model.compile(optimizer="adam", loss="mse")

        early_stop = EarlyStopping(
            monitor="val_loss",
            patience=5,
            restore_best_weights=True
        )

        history = model.fit(
            X_train_seq, y_train_seq,
            epochs=30,
            batch_size=64,
            validation_data=(X_test_seq, y_test_seq),
            callbacks=[early_stop],
            verbose=1
        )

        # Predictions
        y_pred_train = model.predict(X_train_seq, verbose=0).flatten()
        y_pred_test  = model.predict(X_test_seq, verbose=0).flatten()

        rmse_train_lstm = np.sqrt(mean_squared_error(y_train_seq, y_pred_train))
        r2_train_lstm   = r2_score(y_train_seq, y_pred_train)
        rmse_test_lstm  = np.sqrt(mean_squared_error(y_test_seq, y_pred_test))
        r2_test_lstm    = r2_score(y_test_seq, y_pred_test)

        print(f"\n✓ LSTM – Train RMSE: {rmse_train_lstm:.4f}, R²: {r2_train_lstm:.4f}")
        print(f"✓ LSTM –  Test RMSE: {rmse_test_lstm:.4f},  R²: {r2_test_lstm:.4f}")

        # Log metrics
        mlflow.log_metric("rmse_train", rmse_train_lstm)
        mlflow.log_metric("r2_train",   r2_train_lstm)
        mlflow.log_metric("rmse_test",  rmse_test_lstm)
        mlflow.log_metric("r2_test",    r2_test_lstm)

        # Log model hyperparameters
        mlflow.log_param("model_type", "LSTM")
        mlflow.log_param("sequence_length", SEQUENCE_LENGTH)
        mlflow.log_param("lstm_units_1", 64)
        mlflow.log_param("lstm_units_2", 32)
        mlflow.log_param("dropout", 0.2)
        mlflow.log_param("batch_size", 64)
        mlflow.log_param("epochs_planned", 30)
        mlflow.log_param("train_end_year", TRAIN_END_YEAR)

        print(f"✓ MLflow Run ID (LSTM): {mlflow.active_run().info.run_id}")

        # Optional: log model itself
        # mlflow.keras.log_model(model, artifact_path="lstm_model")

        # Plot predictions vs. actuals for first 100 test samples
        plt.figure(figsize=(12, 6))
        plt.plot(y_test_seq[:100], label="Actual", marker="o")
        plt.plot(y_pred_test[:100], label="Predicted", marker="x", alpha=0.7)
        plt.title("LSTM: Actual vs Predicted (First 100 Test Samples)")
        plt.xlabel("Sample Index")
        plt.ylabel("Log Emissions")
        plt.legend()
        plt.grid(True)
        plt.tight_layout()
        plt.show()
else:
    print("\n✗ Not enough sequences generated for LSTM training")
```

```python
def train_final_lstm(
    X_train_data,
    y_train_data,
    X_test_data,
    y_test_data,
    n_features,
    units_1=32,
    units_2=32,
    dropout=0.1,
    sequence_length=4,
    batch_size=32,
    epochs=30
):
    """
    Train LSTM with the best hyperparameters found from sweep.
    """

    with mlflow.start_run(run_name="LSTM_Final_BestConfig"):

        # Build model
        model = Sequential([
            LSTM(units_1, activation="relu",
                 return_sequences=True,
                 input_shape=(sequence_length, n_features)),
            Dropout(dropout),
            LSTM(units_2, activation="relu"),
            Dropout(dropout),
            Dense(1)
        ])
        model.compile(optimizer="adam", loss="mse")

        # Early stopping
        early_stop = EarlyStopping(
            monitor="val_loss",
            patience=5,
            restore_best_weights=True
        )

        # Train
        history = model.fit(
            X_train_data, y_train_data,
            epochs=epochs,
            batch_size=batch_size,
            validation_data=(X_test_data, y_test_data),
            callbacks=[early_stop],
            verbose=1
        )

        # Predictions
        y_pred_train = model.predict(X_train_data).flatten()
        y_pred_test  = model.predict(X_test_data).flatten()

        rmse_train = np.sqrt(mean_squared_error(y_train_data, y_pred_train))
        r2_train   = r2_score(y_train_data, y_pred_train)

        rmse_test = np.sqrt(mean_squared_error(y_test_data, y_pred_test))
        r2_test   = r2_score(y_test_data, y_pred_test)

        # Log metrics
        mlflow.log_metric("rmse_train", rmse_train)
        mlflow.log_metric("r2_train",   r2_train)
        mlflow.log_metric("rmse_test",  rmse_test)
        mlflow.log_metric("r2_test",    r2_test)

        # Log parameters
        mlflow.log_param("units_1", units_1)
        mlflow.log_param("units_2", units_2)
        mlflow.log_param("dropout", dropout)
        mlflow.log_param("sequence_length", sequence_length)
        mlflow.log_param("batch_size", batch_size)
        mlflow.log_param("epochs", epochs)

        print("\n=== FINAL LSTM RESULTS ===")
        print(f"RMSE Train: {rmse_train:.4f}, R² Train: {r2_train:.4f}")
        print(f"RMSE Test : {rmse_test:.4f}, R² Test : {r2_test:.4f}")

        return model, (rmse_train, rmse_test, r2_train, r2_test)
```

```python
X_train_best, y_train_best = create_sequences(
    train_data,
    target_col=target_col,
    group_col="group_key",
    seq_len=4
)
X_test_best, y_test_best = create_sequences(
    test_data,
    target_col=target_col,
    group_col="group_key",
    seq_len=4
)

n_features_best = X_train_best.shape[2]
```

```python
final_lstm_model, final_metrics = train_final_lstm(
    X_train_best,
    y_train_best,
    X_test_best,
    y_test_best,
    n_features=n_features_best
)
```
